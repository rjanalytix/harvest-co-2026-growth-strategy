# 05 · Discounts

Where discount dollars go, and how that compares with where revenue comes from.

Supports deck slides **7, A3**.

![Discount vs revenue](../reports/figures/s7_discount_vs_revenue.png)

```python
import sys
sys.path.insert(0, "../src")
import numpy as np
import pandas as pd
from common import load, customers, weekly_new, CHANNEL_ORDER, LAST_WEEK_IDX
pd.set_option("display.width", 140)
df = load()
```

```python
tot = df.discount_amount.sum()
print(f"Total discount ${tot:,.0f} = {100 * tot / df.gross_revenue.sum():.1f}% of gross revenue (${df.gross_revenue.sum():,.0f})")
fo = df[df.is_first_delivery == 1]
print(f"First orders: {100 * fo.discount_amount.sum() / tot:.1f}% of discount $  ·  repeat orders: {100 * (1 - fo.discount_amount.sum() / tot):.1f}%")
print(f"Average first-order discount (sum of discount / sum of gross): {100 * fo.discount_amount.sum() / fo.gross_revenue.sum():.1f}%")
print(f"First orders: {100 * fo.net_revenue.sum() / df.net_revenue.sum():.1f}% of net revenue")
```

```python
share = (100 * df.groupby("discount_type").discount_amount.sum() / tot).round(1)
print("Share of discount $ by type:", share.sort_values(ascending=False).to_dict())
g = df.groupby("d_year")[["discount_amount", "gross_revenue"]].sum()
print("Discount rate by delivery year (%):", (100 * g.discount_amount / g.gross_revenue).round(1).to_dict())
```
