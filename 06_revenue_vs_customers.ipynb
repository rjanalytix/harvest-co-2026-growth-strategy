# 06 · Revenue vs. customers, year over year

Why revenue kept growing in 2025 while new and active customers both fell. Revenue is by **delivery year**; new customers by **acquisition year**; active = any delivery that year.

Supports deck slides **2, 3, A9**.

![Indexed revenue vs customers](../reports/figures/a9_revenue_vs_customers.png)

```python
import sys
sys.path.insert(0, "../src")
import numpy as np
import pandas as pd
from common import load, customers, weekly_new, CHANNEL_ORDER, LAST_WEEK_IDX
pd.set_option("display.width", 140)
df = load()
```

```python
g = df.groupby("d_year").agg(net_revenue=("net_revenue", "sum"), deliveries=("delivery_id", "count"),
                             active_customers=("customer_id", "nunique"))
g["new_customers"] = df[df.is_first_delivery == 1].groupby("a_year").size()
g["net_rev_yoy_%"] = (g.net_revenue.pct_change() * 100).round(1)
g["active_yoy_%"] = (g.active_customers.pct_change() * 100).round(1)
g["net_rev_per_active"] = (g.net_revenue / g.active_customers).round(0)
g["deliveries_per_active"] = (g.deliveries / g.active_customers).round(1)
g["net_rev_per_delivery"] = (g.net_revenue / g.deliveries).round(2)
g["net_revenue"] = g.net_revenue.round(0)
g
```

```python
ix = (100 * g[["net_revenue", "active_customers", "new_customers"]] / g[["net_revenue", "active_customers", "new_customers"]].iloc[0]).round(0)
print("Indexed to 2023 = 100")
ix
```

Net revenue per active customer rose $427 → $620 (+45%) from 2024 to 2025 while revenue per delivery stayed flat at ~$85: the growth came from order frequency, not price.
